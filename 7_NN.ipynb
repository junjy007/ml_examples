```python



# Calculate model predictions for the grid points
outputs = outputs.reshape(x1.shape)

# Plot the model behavior
plt.figure(figsize=(8, 6))
plt.contourf(x1, x2, outputs, levels=50, cmap='RdBu', alpha=0.7)
plt.colorbar(label='Predicted Probability')

# Add decision boundary (where prediction = 0.5)
plt.contour(x1, x2, outputs, levels=[0.5], colors='black', linewidths=2)
```
