```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
n = 64
N = n*n
J = np.zeros((N, N))
p = 2
for i in range(0, p):
    imfile = f"Data/BW/{i}.webp"
    image = Image.open(imfile).convert('RGB')
    image = image.resize((n, n))
    image = (np.array(image)[:, :, 0] > 128).astype(np.float32)
    image = image[::-1, :]
    # plt.imshow(image)
    # py, px = np.nonzero(1-image)
    s = (image.flatten() * 2.0 - 1.0)
    J += s.reshape([N, 1]).dot(s.reshape(1, N))

# J = J-p/2.0
for i in range(N):
    J[i,i]=0
np.save("hnn_weight.npy", J)

```
